# Python Syntax for Beginners

A step-by-step introduction to Python, building each concept on top of the last.

## 1. Values

Python works with different kinds of information:

```python
10
3.14
type("Hello")
true
```

What this means:

`10` → integer

`3.14` → decimal number (`float`)

`"Hello"` → text (`string`)

`True` → boolean (true/false)

## 2. Variables

Variables give names to values:

```python
age = 20
name = "Mikel"
```

Here, `=` is assignment:

"Store the value 18 under the name `age`."

```python
age
```

```python
name
```

## 3. Operators

Allows you to perform calculations and comparisons

```python
new_age = age + 5
new_age
```

```python
new_age < age
```

## 4. Functions

A function is a named, reusable block of code that performs a specific task:

```python
print("Hello")
print(len("Hello"))
print(round(3.14159, 2))
```

The basic syntax is:

`function(arguments)`

The parentheses hold whatever information the function needs, and commas separate multiple pieces of information.

## 5. Lists

Python can also store multiple values at once:

```python
ages = [18, 19, 20, 21]
```

And access them:

```python
ages[0]
```

Python starts counting positions from `0`, so `ages[0]` gives you the first item in the list.

Tuple: ordered, unchangeable collection. Similar to a list, but once created, its values cannot be changed, added, or removed — (10, 20, 30)

Dictionary: stores values using names (keys) instead of positions — {"country": "UK", "GDP": 3.4}

Set: stores unique values, meaning duplicates are automatically removed — {1, 2, 3}

## 6. Conditions and Indentation

This is where Python's syntax becomes particularly important:

```python
age = 62

if 18 < age < 60:
    print("Adult")
elif age > 60:
    print("elderly")                
else:
    print("teenager")
```

There are three important pieces here:

`if condition:`

&nbsp;&nbsp;&nbsp;&nbsp;`indented code`

The `:` says a block of code is beginning, and indentation tells Python which code belongs to that block.

## 7. Loops

You can repeat an action for every item in a list:

```python
ages = [18, 19, 20]

for age in ages:
    print(age)
```

In plain words:

"For each `age` inside `ages`, print that age."

## Basic Statistics

Python can be used to calculate basic statistics from a set of values.


## 8. Mean

The mean is the average of all values.

It is calculated by adding all the values and dividing by the number of values.

```python
# Read the Excel file and store it as a DataFrame
df2 = pd.read_excel("A01_Spain.xlsx")

# Calculate the mean of A01
df2["A01"].mean()
```

What this means:

`pd.read_excel()` → reads the Excel file

`df2` → stores the data from the file

`df2["A01"]` → selects the A01 column

`.mean()` → calculates its average

## 9. Median

The median is the middle value when all values are arranged from smallest to largest.

If there are two middle values, the median is their average.

```python
# Calculate the median of A01
df2["A01"].median()
```

## 10. Mode

The mode is the value that appears most frequently.

A dataset can have more than one mode if multiple values appear the same number of times.

```python
# Calculate the mode of A01
df2["A01"].mode()
```

## 11. Standard Deviation

The standard deviation measures how spread out the values are around the mean.

A smaller standard deviation means that the values are closer together.

A larger standard deviation means that the values are more spread out.

```python
# Calculate the standard deviation of A01
df2["A01"].std()
```

## 12. Minimum and Maximum

The minimum is the smallest value in the dataset.

The maximum is the largest value in the dataset.

```python
# Find the minimum Bstart value
df2["A01"].min()
```

```python
# Find the maximum Bstart value
df2["A01"].max()
```

## 13. Scatter Plot

A scatter plot shows the relationship between two numerical variables.

Each point represents one observation.

Here, each point represents a year and its corresponding A01 value.

```python
# Import matplotlib and give pyplot the alias "plt"
import matplotlib.pyplot as plt

# Create a scatter plot of Year and Bstart
plt.scatter(df2["Year"], df2["A01"])

# Label the axes
plt.xlabel("Year")
plt.ylabel("A01")

# Display the plot
plt.show()
```

What this means:

`plt.scatter()` → creates a scatter plot

`df["Year"]` → determines the horizontal position of each point

`df["A01"]` → determines the vertical position of each point

`plt.xlabel()` and `plt.ylabel()` → add labels to the axes

`plt.show()` → displays the graph

## 14. Line Plot

A line plot connects observations with lines.

It is useful for showing how a variable changes over time.

Here, the line shows how A01 changed in the United States from 2002 to 2022.

```python
# Create a line plot of Bstart over time
plt.plot(df2["Year"], df2["A01"])

# Label the axes
plt.xlabel("Year")
plt.ylabel("A01")

# Display the plot
plt.show()
```
