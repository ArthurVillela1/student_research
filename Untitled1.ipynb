# Python Syntax for Beginners

A step-by-step introduction to Python, building each concept on top of the last.

## 1. Values

Python works with different kinds of information:

```python
10
3.14
"Hello"
True
```

What this means:

`10` → integer

`3.14` → decimal number (`float`)

`"Hello"` → text (`string`)

`True` → boolean (true/false)

## 2. Variables

Variables give names to values:

```python
age = 18
name = "Mikel"
```

Here, `=` is assignment:

"Store the value 18 under the name `age`."

```python
age
```

```python
name
```

## 3. Operators

Allows you to perform calculations and comparisons

```python
new_age = age + 5
new_age
```

```python
new_age < age
```

## 4. Functions

Functions are commands that perform an operation:

```python
print("Hello")
print(len("Hello"))
print(round(3.14159, 2))
```

The basic syntax is:

`function(arguments)`

The parentheses hold whatever information the function needs, and commas separate multiple pieces of information.

## 5. Lists

Python can also store multiple values at once:

```python
ages = [18, 19, 20, 21]
```

And access them:

```python
ages[0]
```

Python starts counting positions from `0`, so `ages[0]` gives you the first item in the list.

## 6. Conditions and Indentation

This is where Python's syntax becomes particularly important:

```python
age = 18

if age >= 18:
    print("Adult")
```

There are three important pieces here:

`if condition:`

&nbsp;&nbsp;&nbsp;&nbsp;`indented code`

The `:` says a block of code is beginning, and indentation tells Python which code belongs to that block.

## 7. Loops

You can repeat an action for every item in a list:

```python
ages = [18, 19, 20]

for age in ages:
    print(age)
```

In plain words:

"For each `age` inside `ages`, print that age."

```python
# Import pandas and give it the alias "pd"
import pandas as pd
```

```python
# Read the SPSS (.sav) file and store it as a pandas DataFrame
df = pd.read_spss("Data1.sav")

# Export the DataFrame to Excel without the pandas row index
df.to_excel("Data1.xlsx", index=False)

# Display the DataFrame
print(df)
```

```python
# Create an empty dictionary to store the DataFrames
dfs = {}

# Loop through the numbers 1 and 2 (start,stop)
for i in range(1, 3):

    # Create the filename using the current value of i
    filename = f"Data{i}.sav"

    # Read each SPSS file and store its DataFrame in the dictionary
    dfs[i] = pd.read_spss(filename)

# Display the dictionary containing all DataFrames
print(dfs)
```
