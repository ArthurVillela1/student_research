```python
# Import pandas and give it the alias "pd"
import pandas as pd
```

```python
# Read the SPSS (.sav) file and store it as a pandas DataFrame
df = pd.read_spss("Data1.sav")

# Export the DataFrame to Excel without the pandas row index
df.to_excel("Data1.xlsx", index=False)

# Display the DataFrame
print(df)
```

```python
# Create an empty dictionary to store the DataFrames
dfs = {}

# Loop through the numbers 1 and 2 (start,stop)
for i in range(1, 3):

    # Create the filename using the current value of i
    filename = f"Data{i}.sav"

    # Read each SPSS file and store its DataFrame in the dictionary
    dfs[i] = pd.read_spss(filename)

# Display the dictionary containing all DataFrames
print(dfs)
```
